```python
import tensorflow as tf
import numpy as np
from degann.equations.system_ode import SystemODE
from degann.equations.equation_utils import system_ode_from_string, extract_iv
from sklearn.model_selection import train_test_split
import pandas as pd
```

```python
f = "y0**2-2*y0+1 y0(0)=0.5"
f_prep = system_ode_from_string(f)
f_prep
```

```python
f = SystemODE()
f.prepare_equations(1, f_prep)
f.solve((0, 5), 1000)
```

```python
solution_table = pd.DataFrame(f.build_table(), columns=[["time_points", "values"]])
X, y = solution_table["time_points"], solution_table["values"]
```

```python
solution_table.head()
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(X, y, test_size=0.2, random_state=24)
```

```python
from degann.search_algorithms import grid_search
from degann.networks.topology.tf_densenet import TensorflowDenseNet
```

## Стандартная реализация

```python
result_loss, result_epoch, result_loss_name, result_optimizer, result_nn = grid_search(
    input_size=1,
    output_size=1,
    data=(X_train, y_train),
    optimizers=["SGD"],
    loss=["MeanSquaredError"],
    min_epoch=150,
    max_epoch=200,
    epoch_step=50,
    val_data=(X_valid, y_valid),
    nn_min_length=2,
    nn_max_length=3,
    nn_alphabet=["0a", "f8"],
    verbose=True
)
```

```python
from degann.networks.topology.parameter_space import DenseNetParameterSpace
from degann.search_algorithms.grid_search import grid_search_new
```

```python
denseNN_params = DenseNetParameterSpace(
    optimizers=["SGD"],
    loss=["MeanSquaredError"],
    min_epoch=150,
    max_epoch=200,
    epoch_step=50,
    nn_min_length=2,
    nn_max_length=3,
    nn_alphabet=["0a", "f8"]
)
```

```python
result_loss1, result_epoch1, result_loss_name1, result_optimizer1, result_nn1 = grid_search_new(
    input_size=1,
    output_size=1,
    data=(X_train, y_train),
    params=denseNN_params,
    val_data=(X_valid, y_valid),
    verbose=True
)
```
